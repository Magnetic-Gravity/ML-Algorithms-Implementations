# Feature Importance Using Random Forest and Decision Tree

- Dataset Link here - https://www.kaggle.com/datasets/rajmukhiya/digit-recognition-with-pixel-features

```python
import pandas as pd
import seaborn as sns
```

```python
df = pd.read_csv('digit.csv')
df.head()
```

```python
X = df.iloc[:,1:]
y = df.iloc[:,0]
```

```python
sns.heatmap(X.iloc[5].values.reshape(28,28))
```

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier()
rf.fit(X,y)
```

```python
rf.feature_importances_
```

```python
rf.feature_importances_.shape
```

```python
sns.heatmap(rf.feature_importances_.reshape(28,28))
```

# How Feature Importance is Calculated?

```python
from sklearn.datasets import make_classification
from sklearn.tree import DecisionTreeClassifier
```

```python
X,y = make_classification(n_samples=5, n_classes=2, n_features=2, n_informative=2,
                          n_redundant=0, random_state=0)
```

```python
clf = DecisionTreeClassifier()
clf.fit(X,y)
```

```python
from sklearn.tree import plot_tree
plot_tree(clf)
```

```python
clf.feature_importances_
```

```python
from sklearn.ensemble import RandomForestClassifier
rf1 = RandomForestClassifier(n_estimators=2)
rf1.fit(X,y)
```

```python
print(rf1.estimators_[0].feature_importances_)
print(rf1.estimators_[1].feature_importances_)
```

```python
rf1.feature_importances_
```
