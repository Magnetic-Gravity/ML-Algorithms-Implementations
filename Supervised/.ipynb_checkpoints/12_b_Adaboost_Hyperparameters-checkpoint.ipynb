```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from mlxtend.plotting import plot_decision_regions
from sklearn.model_selection import train_test_split
from sklearn.datasets import make_circles
```

```python
np.random.seed(42)
X,y = make_circles(n_samples=500, factor=0.1, noise=0.35, random_state=42)
```

```python
# X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
plt.scatter(X[:,0], X[:,1], c=y)
```

```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.model_selection import cross_val_score

abc = AdaBoostClassifier()

np.mean(cross_val_score(abc, X, y, scoring='accuracy', cv=10))
```

```python
abc.fit(X,y)
```

```python
def plot_decision_boundary(clf):
  plt.figure(figsize=(10,5))
  x_range = np.linspace(X.min(), X.max(), 100)
  xx1, xx2 = np.meshgrid(x_range, x_range)
  y_hat = clf.predict(np.c_[xx1.ravel(), xx2.ravel()])
  y_hat = y_hat.reshape(xx1.shape)
  plt.contourf(xx1, xx2, y_hat, alpha=0.2)
  plt.scatter(X[:,0], X[:,1], c=y, cmap='viridis', alpha=0.7)
  plt.title("Adaboost Classifier")
  plt.show()

plot_decision_boundary(abc)
```

# Tunning No. of Estimators

```python
abc = AdaBoostClassifier(n_estimators=1)
abc.fit(X,y)
plot_decision_boundary(abc)
```

```python
abc = AdaBoostClassifier(n_estimators=100)
abc.fit(X,y)
plot_decision_boundary(abc)
```

```python
abc = AdaBoostClassifier(n_estimators=150)
abc.fit(X,y)
plot_decision_boundary(abc)
```

```python
abc = AdaBoostClassifier(n_estimators=1000)
abc.fit(X,y)
plot_decision_boundary(abc)
```

```python
abc = AdaBoostClassifier(n_estimators=2000)
abc.fit(X,y)
plot_decision_boundary(abc)
```

```python
abc = AdaBoostClassifier(n_estimators=4000)
abc.fit(X,y)
plot_decision_boundary(abc)
```

# Tuning no. of estimators and learning_rate

```python
abc = AdaBoostClassifier(n_estimators=2000, learning_rate=0.1)
abc.fit(X,y)
plot_decision_boundary(abc)
```

# Apply GridSearchCV

```python
from sklearn.model_selection import GridSearchCV

grid = dict()
grid['n_estimators'] = [10, 50, 100, 500]
grid['learning_rate'] = [0.0001, 0.001, 0.01, 0.1, 1.0]
grid['algorithm'] = ['SAMME', 'SAMME.R']

grid_search = GridSearchCV(estimator=AdaBoostClassifier(), param_grid=grid, n_jobs=-1, cv=10, scoring='accuracy')

#execute the grid search
grid_result = grid_search.fit(X,y)

#summarize the best score and configuration
print("Best: %f using %s" % (grid_result.best_score_, grid_result.best_params_))
```
