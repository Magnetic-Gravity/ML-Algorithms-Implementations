```python
data['play'].value_counts()
```

```python
PYes = 9/14
PNo = 5/14
print(PYes)
print(PNo)
```

```python
# Outlook
pd.crosstab(data['outlook'],data['play'])
```

```python
POvercastNo = 0
PRainyNo = 2/5
PSunnyNo = 3/5

POvercastYes = 4/9
PRainyYes = 3/9
PSunnyYes = 2/9
```

```python
# temp
pd.crosstab(data['temp'],data['play'])
```

```python
PCoolNo = 1/5
PHotNo = 2/5
PMildNo = 2/5

PCoolYes = 3/9
PHotYes = 2/9
PMildYes = 4/9
```

```python
# Humidity
pd.crosstab(data['humidity'],data['play'])
```

```python
PHighNo=4/5
PNormalNo=1/5

PHighYes=3/9
PNormalYes=6/9
```

```python
# Wind
pd.crosstab(data['windy'],data['play'])
```

```python
# False=Weak, True=Strong
PWeakNo = 2/5
PStrongNo = 3/5

PWeakYes = 6/9
PStrongYes = 3/9
```

```python
# Problem 1

# Outlook=sunny, Temp=Hot, Humidity=High, Wind=Weak
# Play or No Play?

# Solution -

# P(Yes / Sunny,Hot,High,Weak) = P(Sunny/Yes) * P(Hot/Yes) * P(High/Yes) * P(Weak/Yes) * P(Yes)
# P(No / Sunny,Hot,High,Weak) = P(Sunny/No) * P(Hot/No) * P(High/No) * P(Weak/No) * P(No)

#compare and decide using the maximum a posteriori rule
```

```python
Play_Yes = PSunnyYes * PHotYes * PHighYes * PWeakYes * PYes
print(Play_Yes)
```

```python
No_Play = PSunnyNo * PHotNo * PHighNo * PWeakNo * PNo
print(No_Play)
```

```python
# Hence No play in Problem 1
```

```python
# Problem 2

# Outlook=Overcast, Temp=Cool, Humidity=Normal, Wind=Weak
# Play or No Play?

# Solution -

# P(Yes / Overcast,Cool,Normal,Weak) = P(Overcast/Yes) * P(Cool/Yes) * P(Normal/Yes) * P(Weak/Yes) * PYes
# P(No / Overcast,Cool,Normal,Weak) = P(Overcast/No) * P(Cool/No) * P(Normal/No) * P(Weak/No) * PNo

#compare and decide using the maximum a posteriori rule
```

```python
Play_Yes_day = POvercastYes * PCoolYes * PNormalYes * PWeakYes * PYes
print(Play_Yes_day)
```

```python
No_Play_day = POvercastNo * PCoolNo * PNormalNo * PWeakNo * PNo
print(No_Play_day)
```

```python
# Hence He will Play on Problem 2
```
