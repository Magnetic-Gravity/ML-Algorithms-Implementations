# Voting Ensemble Regression

```python
from sklearn.datasets import fetch_california_housing
housing = fetch_california_housing()
X,y = housing.data, housing.target
```

```python
X.shape,y.shape
```

```python
X[0:2], y[0:2] # print only two rows 
```

```python
import numpy as np
from sklearn.linear_model import LinearRegression
from sklearn.tree import DecisionTreeRegressor
from sklearn.svm import SVR
from sklearn.model_selection import cross_val_score
```

```python
lr = LinearRegression()
dt = DecisionTreeRegressor()
svr = SVR()
```

```python
estimators = [('lr',lr),('dt',dt),('svr',svr)]
```

```python
for estimator in estimators:
    scores = cross_val_score(estimator[1], X, y, cv=10, scoring='r2')
    print(estimator[0], np.round(np.mean(scores),2))
```

```python
from sklearn.ensemble import VotingRegressor
```

```python
vr_hard = VotingRegressor(estimators) # by default voting='hard'
scores = cross_val_score(vr_hard, X, y, cv=10, scoring='r2')
print("Hard Voting Regressor : ", np.round(np.mean(scores),2))
```

```python
vr_soft = VotingRegressor(estimators=estimators,voting='soft') 
scores = cross_val_score(vr_soft, X, y, cv=10, scoring='r2')
print("Soft Voting Regressor : ", np.round(np.mean(scores),2))
```

```python
for i in range(1,4):
    for j in range(1,4):
        for k in range(1,4):
            vr = VotingRegressor(estimators, weights=[i,j,k]) 
            scores = cross_val_score(vr, X, y, cv=10, scoring='r2')
            print("for i={},j={},k={}".format(i,j,k), np.round(np.mean(scores),2))
```

# Using Same Algorithm

```python
dt1 = DecisionTreeRegressor(max_depth=1)
dt2 = DecisionTreeRegressor(max_depth=3)
dt3 = DecisionTreeRegressor(max_depth=5)
dt4 = DecisionTreeRegressor(max_depth=7)
dt5 = DecisionTreeRegressor(max_depth=None)
```

```python
estimators = [('dt1',dt1),('dt2',dt2),('dt3',dt3),('dt4',dt4),('dt5',dt5)]
```

```python
for estimator in estimators:
    scores = cross_val_score(estimator[1], X, y, cv=10, scoring='r2')
    print(estimator[0], np.round(np.mean(scores),2))
```
