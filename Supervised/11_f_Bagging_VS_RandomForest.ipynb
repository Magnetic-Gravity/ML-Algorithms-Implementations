## Difference b/w Bagging using Decision Tree and Random Forest Using Decision Tree

- 1. Bagging takes different base_estimator(DT, SVM, KNN, LR) but Random Forest take only DT by default

- 2. Bagging samples data based on Tree level column sampling but Random Forest samples data based on Node level column sampling

```python
import numpy as np
import pandas as pd
from sklearn.datasets import make_classification
from sklearn.ensemble import BaggingClassifier, RandomForestClassifier
from sklearn.tree import plot_tree
import matplotlib.pyplot as plt
```

```python
X,y = make_classification(n_features=5, n_redundant=0, n_informative=5, n_clusters_per_class=1)
```

```python
df = pd.DataFrame(X, columns=['col1','col2','col3','col4','col5'])
df['target'] = y
print(df.shape)
df.head()
```

# Bagging Classifier

- Tree Level Column Sampling(Suppose you got two columns 3,4 then the whole tree will be split using only these two columns)

```python
bag = BaggingClassifier(max_features=2)
bag.fit(df.iloc[:,:5],df.iloc[:,-1])
```

```python
plt.figure(figsize=(18,12))
plot_tree(bag.estimators_[0])
plt.show()
```

# Random Forest Classifier
- Node Level Column Sampling(Suppose you got randomly two columns 3,4 for root node one of them will be selected for root node but in next left or right node you will randomly select from all columns)

```python
rf = RandomForestClassifier(max_features=2)
rf.fit(df.iloc[:,:5],df.iloc[:,-1])
```

```python
plt.figure(figsize=(18,12))
plot_tree(rf.estimators_[4])
plt.show()
```
